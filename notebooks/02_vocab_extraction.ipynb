```python
import os
import numpy as np
import pandas as pd
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# ----------------------------
# Configuration
# ----------------------------
MODEL_NAME = "facebook/opt-2.7b"
HF_TOKEN = os.getenv("HF_TOKEN")  # Make sure HF_TOKEN is set

print(HF_TOKEN) 
tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    token=HF_TOKEN
)

# ----------------------------
# Load model
# ----------------------------
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    token=HF_TOKEN,
    torch_dtype=torch.float32,
    device_map="auto"      # Change to "auto" if using GPU
)


embedding_matrix = (
    model.get_input_embeddings()
         .weight
         .detach()
         .cpu()
         .numpy()
)

print("Embedding matrix shape:", embedding_matrix.shape)
# Expected: (50272, 2560)

# ----------------------------
# Extract vocabulary
# ----------------------------
vocab = tokenizer.get_vocab()

# Convert token->id to id->token
id_to_token = {idx: token for token, idx in vocab.items()}

tokens = [id_to_token[i] for i in range(len(id_to_token))]

# ----------------------------
# Save vocabulary
# ----------------------------
vocab_df = pd.DataFrame({
    "token_id": range(len(tokens)),
    "token": tokens
})

vocab_df.to_csv("opt27_vocab.csv", index=False)


np.save("opt27_embeddings.npy", embedding_matrix)

print("Vocabulary saved to: opt27_vocab.csv")
print("Embeddings saved to: opt27_embeddings.npy")
```
