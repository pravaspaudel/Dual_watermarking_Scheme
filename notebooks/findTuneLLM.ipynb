```python
import json

records = []
per_file_counts = {}

for fname in dataset_filenames:
    with open(fname, "r", encoding="utf-8") as f:
        content = f.read().strip()

    if content.startswith("["):
        file_records = json.loads(content)
    else:
        file_records = [json.loads(line) for line in content.splitlines() if line.strip()]

    per_file_counts[fname] = len(file_records)
    records.extend(file_records)

print("Records per file:")
for fname, count in per_file_counts.items():
    print(f"  {fname}: {count}")
print(f"\nTotal combined records: {len(records)}")

# Re-run the same validation checks as before
required_fields = {"topic", "difficulty", "question"}
missing_field_count = sum(1 for r in records if not required_fields.issubset(r.keys()))
empty_field_count = sum(
    1 for r in records
    if required_fields.issubset(r.keys())
    and (not r["topic"].strip() or not r["difficulty"].strip() or not r["question"].strip())
)
difficulties = set(r.get("difficulty", "").strip().lower() for r in records)

print(f"\nRecords missing required fields: {missing_field_count}")
print(f"Records with an empty required field: {empty_field_count}")
print(f"Unique difficulty values found: {difficulties}")

# Check for exact duplicate questions across files (common when combining datasets)
questions_seen = [r["question"].strip() for r in records]
duplicate_count = len(questions_seen) - len(set(questions_seen))
print(f"Duplicate questions found: {duplicate_count}")
```

```python
from transformers import AutoTokenizer

MODEL_NAME = "facebook/opt-2.7b"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# Define the prompt template.
# IMPORTANT: this exact template will be reused at inference time.
def build_prompt(topic, difficulty):
    return f"### Input\nTopic: {topic}\nDifficulty: {difficulty}\n\n### Output\n"

def build_full_example(topic, difficulty, question):
    prompt = build_prompt(topic, difficulty)
    full_text = prompt + question + tokenizer.eos_token
    return prompt, full_text

# Test it on the first record
sample = records[0]
prompt, full_text = build_full_example(sample["topic"], sample["difficulty"], sample["question"])
print("--- PROMPT ---")
print(prompt)
print("--- FULL TRAINING TEXT ---")
print(full_text)
```

```python
import numpy as np

prompt_lengths = []
full_lengths = []

for r in records:
    prompt, full_text = build_full_example(r["topic"], r["difficulty"], r["question"])
    prompt_ids = tokenizer(prompt, add_special_tokens=True)["input_ids"]
    full_ids = tokenizer(full_text, add_special_tokens=True)["input_ids"]
    prompt_lengths.append(len(prompt_ids))
    full_lengths.append(len(full_ids))

prompt_lengths = np.array(prompt_lengths)
full_lengths = np.array(full_lengths)

print("=== Prompt (input) token lengths ===")
print(f"min: {prompt_lengths.min()}, max: {prompt_lengths.max()}, mean: {prompt_lengths.mean():.1f}")

print("\n=== Full (prompt + question) token lengths ===")
print(f"min: {full_lengths.min()}, max: {full_lengths.max()}, mean: {full_lengths.mean():.1f}")
print(f"median: {np.median(full_lengths):.1f}, 95th percentile: {np.percentile(full_lengths, 95):.1f}, 99th percentile: {np.percentile(full_lengths, 99):.1f}")
```

```python
from datasets import Dataset

MAX_SEQ_LENGTH = 128

def tokenize_example(r):
    prompt, full_text = build_full_example(r["topic"], r["difficulty"], r["question"])

    prompt_ids = tokenizer(prompt, add_special_tokens=True)["input_ids"]
    full_ids = tokenizer(
        full_text,
        add_special_tokens=True,
        truncation=True,
        max_length=MAX_SEQ_LENGTH,
    )["input_ids"]

    labels = full_ids.copy()
    # Mask out the prompt portion of the labels so loss only applies to the question
    prompt_len = min(len(prompt_ids), len(full_ids))
    for i in range(prompt_len):
        labels[i] = -100

    return {
        "input_ids": full_ids,
        "attention_mask": [1] * len(full_ids),
        "labels": labels,
    }

tokenized_records = [tokenize_example(r) for r in records]

hf_dataset = Dataset.from_list(tokenized_records)
print(hf_dataset)
print("\n--- Example tokenized record ---")
example = hf_dataset[0]
print("input_ids length:", len(example["input_ids"]))
print("labels length:", len(example["labels"]))

# Sanity check: decode only the non-masked (question) part of labels
non_masked_ids = [t for t in example["labels"] if t != -100]
print("\nDecoded target (should be ONLY the question, no input template):")
print(tokenizer.decode(non_masked_ids))

# Sanity check: confirm masked count matches prompt length
masked_count = sum(1 for t in example["labels"] if t == -100)
print(f"\nMasked (prompt) tokens: {masked_count}")
print(f"Unmasked (question) tokens: {len(example['labels']) - masked_count}")
```

```python
from transformers import AutoModelForCausalLM
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training
import torch

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=bnb_config,
    device_map="auto",
)

model = prepare_model_for_kbit_training(model)

lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=["q_proj", "k_proj", "v_proj", "out_proj"],
)

model = get_peft_model(model, lora_config)

model.print_trainable_parameters()
```

Forward-pass sanity check

```python
import torch

# Take a small batch of 4 examples
sample_batch = [hf_dataset[i] for i in range(4)]

max_len_in_batch = max(len(ex["input_ids"]) for ex in sample_batch)

def pad_batch(batch, pad_token_id, max_len):
    input_ids, attention_mask, labels = [], [], []
    for ex in batch:
        pad_len = max_len - len(ex["input_ids"])
        input_ids.append(ex["input_ids"] + [pad_token_id] * pad_len)
        attention_mask.append(ex["attention_mask"] + [0] * pad_len)
        labels.append(ex["labels"] + [-100] * pad_len)
    return {
        "input_ids": torch.tensor(input_ids),
        "attention_mask": torch.tensor(attention_mask),
        "labels": torch.tensor(labels),
    }

pad_token_id = tokenizer.pad_token_id if tokenizer.pad_token_id is not None else tokenizer.eos_token_id
batch = pad_batch(sample_batch, pad_token_id, max_len_in_batch)

# Move to GPU
batch = {k: v.to(model.device) for k, v in batch.items()}

model.train()
outputs = model(**batch)

print("Loss:", outputs.loss.item())
print("Logits shape:", outputs.logits.shape)

# Confirm gradients flow into LoRA params
outputs.loss.backward()

grad_found = False
for name, param in model.named_parameters():
    if param.requires_grad and param.grad is not None and param.grad.abs().sum() > 0:
        grad_found = True
        print(f"Gradient confirmed flowing into: {name}")
        break

if not grad_found:
    print("WARNING: No gradient detected in trainable parameters!")

# Clear gradients so this test doesn't affect the real training run
model.zero_grad()

print(f"\nGPU memory allocated after forward+backward: {torch.cuda.memory_allocated() / 1e9:.2f} GB")
```

```python
from transformers import DataCollatorForSeq2Seq, TrainingArguments

data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    padding=True,          # dynamic padding per batch
    label_pad_token_id=-100,
)

training_args = TrainingArguments(
    output_dir="./opt27b-lora-checkpoints",
    per_device_train_batch_size=8,
    gradient_accumulation_steps=1,
    num_train_epochs=3,
    learning_rate=2e-4,
    fp16=True,
    optim="paged_adamw_8bit",
    logging_steps=5,
    save_strategy="no",       # we'll save the final adapter manually after training
    report_to="none",
    warmup_steps=3,
    lr_scheduler_type="cosine",
)

print("TrainingArguments configured successfully.")
steps_per_epoch = len(hf_dataset) // training_args.per_device_train_batch_size
total_steps = steps_per_epoch * training_args.num_train_epochs
print(f"Steps per epoch: {steps_per_epoch}")
print(f"Total training steps: {total_steps}")
```

```python
from transformers import Trainer
import time

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=hf_dataset,
    data_collator=data_collator,
)

start_time = time.time()
train_result = trainer.train()
end_time = time.time()

print(f"\nTraining completed in {(end_time - start_time) / 60:.2f} minutes")
print(f"Final training loss: {train_result.training_loss:.4f}")
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
from peft import PeftModel
from transformers import BitsAndBytesConfig
import torch

MODEL_NAME = "facebook/opt-2.7b"
ADAPTER_DIR = "./opt27b-question-generator-lora"

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)

base_model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=bnb_config,
    device_map="auto",
)

inference_model = PeftModel.from_pretrained(base_model, ADAPTER_DIR)
inference_model.eval()

inference_tokenizer = AutoTokenizer.from_pretrained(ADAPTER_DIR)

def generate_question(topic, difficulty, max_new_tokens=80):
    prompt = f"### Input\nTopic: {topic}\nDifficulty: {difficulty}\n\n### Output\n"
    inputs = inference_tokenizer(prompt, return_tensors="pt").to(inference_model.device)

    with torch.no_grad():
        output_ids = inference_model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=0.7,
            top_p=0.9,
            repetition_penalty=1.15,
            eos_token_id=inference_tokenizer.eos_token_id,
            pad_token_id=inference_tokenizer.pad_token_id or inference_tokenizer.eos_token_id,
        )

    generated_ids = output_ids[0][inputs["input_ids"].shape[1]:]
    question = inference_tokenizer.decode(generated_ids, skip_special_tokens=True)
    return question.strip()

# Test with the example from your original prompt
result = generate_question("Rational Agent Framework", "medium")
print("Generated question:")
print(result)
```
