```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

model_name = "facebook/opt-2.7b"

# Load the tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Load the model
# Using torch.bfloat16 for memory efficiency if available, otherwise float16
if torch.cuda.is_available():
    model = AutoModelForCausalLM.from_pretrained(model_name, torch_dtype=torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16)
    model.to('cuda')
else:
    print("CUDA not available. Loading model on CPU with float32. This might be slow.")
    model = AutoModelForCausalLM.from_pretrained(model_name)


print(f"Tokenizer and model '{model_name}' loaded successfully.")
```

```python
# Encode the prompt
inputs = tokenizer(prompt, return_tensors="pt")

# Move inputs to GPU if model is on GPU
if torch.cuda.is_available():
    inputs = {k: v.to('cuda') for k, v in inputs.items()}

# Get model outputs (including logits)
with torch.no_grad(): # Disable gradient calculation for inference
    outputs = model(**inputs)

# Extract raw logits
# The logits are usually the first element in the model's output tuple
# and correspond to the vocabulary size for each token position.
raw_logits = outputs.logits

print("Raw logits (scores before softmax) for the prompt tokens:")
print(f"Shape of raw logits: {raw_logits.shape}")
print(raw_logits)
```

```python
# Extract token IDs from the inputs
token_ids = inputs['input_ids'][0].tolist()



tokens = tokenizer.convert_ids_to_tokens(token_ids)

print("Token | Token ID")
print("------|---------")
for i in range(len(tokens)):
    print(f"{tokens[i]:<5} | {token_ids[i]}")
```

```python
decoded_text = tokenizer.decode(token_ids)
print(f"\nDecoded text from token IDs: {decoded_text}")
```

```python

# Get the predicted token IDs by taking the argmax of the raw logits
# raw_logits has shape [batch_size, sequence_length, vocab_size]
# We want the token with the highest logit for each position in the sequence
predicted_token_ids = torch.argmax(raw_logits, dim=-1)

# The output for a single batch item will be in predicted_token_ids[0]
# Decode the predicted token IDs back into text
generated_text = tokenizer.decode(predicted_token_ids[0], skip_special_tokens=True)

print("\nText generated from raw logits (most probable tokens):")
print(generated_text)
```

```python

prompt_token_ids = inputs['input_ids'][0].tolist()

prompt_relevant_greenlist = topic_greenlists_df[topic_greenlists_df['token_id'].isin(prompt_token_ids)]

topic_scores = prompt_relevant_greenlist.groupby('topic')['similarity'].sum().reset_index()

best_topic = topic_scores.loc[topic_scores['similarity'].idxmax()]

print("\nTopic Similarity Scores for the Prompt:")
display(topic_scores.sort_values(by='similarity', ascending=False))

print(f"\nThe identified topic for the prompt is: {best_topic['topic']} with a total similarity score of {best_topic['similarity']:.4f}")
```

```python


# 1. Define the delta value
delta = 2.0  # The bias value to add to greenlist tokens

# 2. Identify the greenlist token IDs for the identified topic ('sports')
# We use the topic identified in the previous step
target_topic = best_topic['topic']
greenlist_ids = topic_greenlists_df[topic_greenlists_df['topic'] == target_topic]['token_id'].unique().tolist()

# 3. Create a copy of the raw logits to modify
watermarked_logits = raw_logits.clone()

# 4. Add delta to the logits of the greenlist tokens at each position
# raw_logits shape is [batch_size, sequence_length, vocab_size]
for token_id in greenlist_ids:
    if token_id < watermarked_logits.shape[-1]:
        watermarked_logits[:, :, token_id] += delta

print(f"Applied delta={delta} to {len(greenlist_ids)} tokens belonging to the '{target_topic}' topic.")

# 5. Let's see the effect by getting the new predicted tokens
watermarked_predicted_ids = torch.argmax(watermarked_logits, dim=-1)
watermarked_text = tokenizer.decode(watermarked_predicted_ids[0], skip_special_tokens=True)

print("\nText generated from watermarked logits:")
print(watermarked_text)
```

```python
def apply_watermark(prompt, delta, model, tokenizer, greenlist_df):
    """
    Identifies the best topic for a prompt and applies a logit bias (delta)
    to the greenlist tokens of that topic.
    """
    # 1. Tokenize and get raw logits
    inputs = tokenizer(prompt, return_tensors="pt")
    if torch.cuda.is_available():
        inputs = {k: v.to('cuda') for k, v in inputs.items()}

    with torch.no_grad():
        outputs = model(**inputs)
    logits = outputs.logits

    # 2. Identify the best topic
    prompt_ids = inputs['input_ids'][0].tolist()
    relevant_greenlist = greenlist_df[greenlist_df['token_id'].isin(prompt_ids)]

    if relevant_greenlist.empty:
        print("No matching greenlist tokens found for topic identification. Using default bias.")
        return tokenizer.decode(torch.argmax(logits, dim=-1)[0], skip_special_tokens=True), "None"

    topic_scores = relevant_greenlist.groupby('topic')['similarity'].sum()
    best_topic_name = topic_scores.idxmax()

    # 3. Apply Delta to Greenlist tokens
    watermarked_logits = logits.clone()
    target_greenlist_ids = greenlist_df[greenlist_df['topic'] == best_topic_name]['token_id'].unique()

    for tid in target_greenlist_ids:
        if tid < watermarked_logits.shape[-1]:
            watermarked_logits[:, :, tid] += delta

    # 4. Decode the result
    predicted_ids = torch.argmax(watermarked_logits, dim=-1)
    text = tokenizer.decode(predicted_ids[0], skip_special_tokens=True)

    return text, best_topic_name

# Quick test with the new function
test_prompt = "the athlete won a gold medal at the stadium"
result_text, detected_topic = apply_watermark(test_prompt, delta=3.0, model=model, tokenizer=tokenizer, greenlist_df=topic_greenlists_df)

print(f"Detected Topic: {detected_topic}")
print(f"Watermarked Output: {result_text}")
```

```python
# Updated with longer prompts for better autoregressive context
demo_prompts = [
    "The professional athlete won a prestigious gold medal at the national stadium after years of training in various sports disciplines.",
    "The experienced doctor prescribed a new type of medicine to the patient to treat the infection and improve overall health.",
    "The software engineer used the latest technology and computer programming tools to build a highly scalable web application."
]

delta_value = 4.0  # Increased slightly to make the effect more distinct
results = []

for p in demo_prompts:
    # Get original (unwatermarked) output
    original, _ = apply_watermark(p, delta=0.0, model=model, tokenizer=tokenizer, greenlist_df=topic_greenlists_df)

    # Get watermarked output
    watermarked, topic = apply_watermark(p, delta=delta_value, model=model, tokenizer=tokenizer, greenlist_df=topic_greenlists_df)

    results.append({
        "Prompt": p,
        "Detected Topic": topic,
        "Original Output": original,
        "Watermarked Output": watermarked
    })

# Display the results table
comparison_df = pd.DataFrame(results)
display(comparison_df)
```
